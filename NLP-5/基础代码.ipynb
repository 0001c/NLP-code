```python
import warnings
warnings.filterwarnings("ignore")
```

```python
import tensorflow as tf
import tensorflow.compat.v1 as tf1
tf.compat.v1.disable_v2_behavior() # 1.x 兼容模式
# TF 2.16+ disable_v2_behavior 不再导出 Session 等到顶层, 手动补上
tf.Session = tf1.Session
tf.placeholder = tf1.placeholder
tf.global_variables_initializer = tf1.global_variables_initializer
tf.Variable = tf1.Variable
tf.train = tf1.train
tf.layers = tf1.layers
tf.losses = tf1.losses
```

### 简单的示例

```python
ht = tf.constant('hello,tf', name='hello')
print(ht)
```

```python
sess = tf.Session()
```

```python
ht_value = sess.run(ht)
print(ht_value)
```

```python
# 防止内存泄漏
sess.close()
```

```python
# 使用with语句自动关闭会话
with tf.Session() as sess:
    ht_value = sess.run(ht)
    print(ht_value)
```

### 标准模式

```python
# 创建一个graph对象
graph_t = tf.Graph()

# 将graph对象作为默认的图
with graph_t.as_default():
    ht = tf.constant('hello,tf', name='hello')

# 创建session对象
with tf.Session(graph=graph_t) as sess:
    ht_value = sess.run(ht)
    print(ht_value)
```

### 简单计算

```python
graph_t = tf.Graph()

with graph_t.as_default():
    a = tf.constant(1, name='a')
    b = tf.constant(2, name='b')
    c = a + b
    print(c)
    
# 创建session对象
with tf.Session(graph=graph_t) as sess:
    c_value = sess.run(c)
    print(c_value)
```

```python
# 看下a,b,c的值
with tf.Session(graph=graph_t) as sess:
    a_value,b_value,c_value = sess.run([a,b,c])
    print(a_value, b_value, c_value)
    
```

### placeholder

```python
a = tf.placeholder(tf.float32)
print(a)
```

```python
graph_t = tf.Graph()
with graph_t.as_default():
    a = tf.placeholder(tf.float32)

with tf.Session(graph=graph_t) as sess:
    a_value = sess.run(a, feed_dict={a:1.0})
    print(a_value)
```

```python
graph_t = tf.Graph()
with graph_t.as_default():
    a = tf.placeholder(tf.float32)
    b = tf.placeholder(tf.float32)
    c = a + b
    print(c)
    
with tf.Session(graph=graph_t) as sess:
    c_value = sess.run(c, feed_dict={a:1.0, b:2.0})
    print(c_value)
```

#### 练习

建立一个c=（a+2b）*b的计算图，传入任意的a,b值，计算c的值

```python
graph_t = tf.Graph()
with graph_t.as_default():
    a = tf.placeholder(tf.float32)
    b = tf.placeholder(tf.float32)
    c = (a + 2*b) * b

with tf.Session(graph=graph_t) as sess:
    c_value = sess.run(c, feed_dict={a:1.0, b:2.0})
    print(c_value)
```

### variable

```python
a = tf.Variable(1.0,dtype=tf.float32)
print(a)

```

```python
graph_t = tf.Graph()
with graph_t.as_default():
    a = tf.Variable(1.0,dtype=tf.float32)
    b = tf.Variable(2.0,dtype=tf.float32)
    c = a + b

with tf.Session(graph=graph_t) as sess:
    c_value = sess.run(c)
    print(c_value)
```

```python
graph_t = tf.Graph()
with graph_t.as_default():
    a = tf.Variable(1.0,dtype=tf.float32)
    b = tf.Variable(2.0,dtype=tf.float32)
    c = a + b

with tf.Session(graph=graph_t) as sess:
    sess.run(tf.global_variables_initializer())  # 初始化变量
    c_value = sess.run(c)
    print(c_value)
```
