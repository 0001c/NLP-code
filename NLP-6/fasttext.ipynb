# fasttext 模型

## 简单用法

```python
def train_unsupervised(input, model="skipgram", lr=0.05, dim=100,
        ws=5, epoch=5, minCount=5,
        minCountLabel=0, minn=3,
        maxn=6, neg=5, wordNgrams=1,
        loss="ns", bucket=2000000,
        thread=12, lrUpdateRate=100,
        t=1e-4, label="__label__",
        verbose=2, pretrainedVectors="")

* input : 训练文件路径，文件每一行代表一篇文档，文档是已经经过分词。
* model : 无监督训练使用的模型 {cbow 和 skip-gram} 默认为 skip-gram
* lr : 学习率 默认 0.05
* dim : 词向量维度 默认为 100
* ws : (windows size) 窗口大小 默认为 5
* epoch : 默认训练五轮
* minCount : 最小词数 默认为 5 当数据集小要记得调整
* minCountLabel : 最小标签数 默认为 0 因为是无监督。
* minn : 最小 char 级别的 3-gram (subword)
* maxn : 最大 char 级别的 6-gram (subword)
* neg : 负例采样个数 默认为 5
* wordNgrams : 最大的词 n-gram 长度 默认为 1
* loss : 损失函数 {ns,hs,softmax,ova} 默认是 ns 负采样，hs 是分层 softmax   


````python
import fasttext
````

````python
filepath = 'data/mb_train.txt'
model = fasttext.train_unsupervised(filepath, model='skipgram', dim=200, ws=5)
model.save_model('models/mb_fasttext.model')
````

````python
model = fasttext.load_model('models/mb_fasttext.model')
````

````python
# 获取单词的fasttext向量
print(model.get_word_vector('安卓'))
````

````python
# 获取训练语料中不存在的单词的向量
print(model.get_word_vector('nihao'))
````
